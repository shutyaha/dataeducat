```python
n = int(input())
cnt = 0

while n > 0:
    while n >= 5:
        while n >= 10:
            while n >= 25:
                cnt += 1
                n -= 25
            cnt += 1
            n -= 10
        cnt += 1
        n -= 5
    cnt += 1
    n -= 1
print(cnt)
```

```python
#28n+30k+31m=365

for n in range(1, 14):
    for k in range(1, 13):
        for m in range(1, 12):
            if 28 * n + 30 * k + 31 * m == 365:
                print(n, k, m)
```
